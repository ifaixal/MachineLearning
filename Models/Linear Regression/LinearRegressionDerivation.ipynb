```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
class LinearRegression:
    def __init__(self):
        self.m = None
        self.b = None

    def fit(self, X_train, y_train):
        # Finding M
        # Calculating M = summation from i to n ((x_i - x.mean) (y_i - y.mean))/(summation from i to n (x_i - x.mean)^2)
        X_mean = X_train.mean()
        Y_mean = y_train.mean()
        nominator = 0
        for i in range(len(X_train)):
            nominator += (X_train.iloc[i] - X_mean) * (y_train.iloc[i] - Y_mean)
        denominator = 0
        for i in range(len(X_train)):
            denominator += ((X_train.iloc[i] - X_mean)**2)
        
        self.m = nominator / denominator

        # Calculating B
        self.b = Y_mean - (m * X_mean)

        print(self.m, " = m")
        print(self.b, ' = b')
    
    def predict(self, X_test):
        return (self.m * X_test) - self.b
```
